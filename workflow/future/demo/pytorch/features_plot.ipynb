```python
import panel.widgets as pnw
import xarray as xa
from argparse import Namespace
import tmodel
import hvplot as hv
import numpy as np
import pandas as pd
import hvplot.pandas  # noqa
import panel as pn
hv.extension('bokeh')

signal_index=2
feature_type=0
args = Namespace( nfeatures=64, minp_factor=1.0 )
```

```python
data=tmodel.get_demo_data()
T: np.ndarray = data['times'][signal_index]
F: np.ndarray = tmodel.get_features( T, feature_type, args )
```

```python
df = pd.DataFrame({'x': T})
def feature_plot(feature):
	df['f'] = F[:,feature]
	return df.hvplot.line( x='x', y='f', title='Features', frame_width=1000 )

feature_slider = pn.widgets.IntSlider( name='Feature', start=0, end=F.shape[1]-1, value=0, width=200 )
plot = pn.bind(feature_plot, feature=feature_slider)

pn.Column(feature_slider, plot)
```
