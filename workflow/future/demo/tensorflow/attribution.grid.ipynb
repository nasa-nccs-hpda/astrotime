```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np, os, time
import xarray as xa
from argparse import Namespace
import tmodel

signal_index = 2
iFT = 0

args: Namespace = tmodel.load_args(signal_index, 0)
att_path = tmodel.attribution_path( args, signal_index )
att_ds: xa.Dataset = xa.open_dataset( att_path )

data=tmodel.get_demo_data()
T: np.ndarray = data['times'][signal_index]
att_var = att_ds.data_vars[f"AF{iFT}"].values
ATT0: np.ndarray = att_var[0]
ATTf: np.ndarray = att_var[1:]
```

```python
fig, axs = plt.subplots(3, 3, figsize=(20,50))
for ir in range(3):
	for ic in range(3):
		ip = ir*3+ic
		ax = axs[ir,ic]
		ax.plot( T, ATT0, label='Target', color='green', linewidth=1, alpha=0.7 )
		ax.plot( T, ATTf[ip], label=f'Feature {ip} masked', color='blue',  linewidth=1,  alpha=0.7 )
		ax.set_title(f"Masked Attribution: Feature {ip}")
plt.tight_layout()
plt.show()
```
