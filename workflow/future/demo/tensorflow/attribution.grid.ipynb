```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np, os, time
import xarray as xa
from argparse import Namespace
import tmodel

signal_index = 2
iFT = 0

args: Namespace = tmodel.load_args(signal_index, 0)
att_path = tmodel.attribution_path( args, signal_index )
att_ds: xa.Dataset = xa.open_dataset( att_path )

data=tmodel.get_demo_data()
att_var = att_ds.data_vars[f"AF{iFT}"].values
ATT0: np.ndarray = att_var[0]
ATTf: np.ndarray = att_var[1:]
T: np.ndarray = data['times'][signal_index][:ATT0.size]
```

```python
fig, axs = plt.subplots(3, 3, figsize=(15,7))
plt.title("Feature Attribution: Model results with a single feature masked (masking results: green)")
for ir in range(3):
	for ic in range(3):
		ip = ir*3+ic
		ax = axs[ir,ic]
		ax.plot( T, ATT0, color='green', linewidth=1, alpha=0.5 )
		ax.plot( T, ATTf[ip], color='blue',  linewidth=1,  alpha=0.5 )
		ax.set_title(f"Results with feature {ip} Masked")
fig.tight_layout()
plt.show()
```
