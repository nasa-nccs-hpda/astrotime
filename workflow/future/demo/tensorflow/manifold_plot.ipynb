```python
from argparse import Namespace
import tmodel
import numpy as np
import holoviews as hv

hv.extension('plotly')
use_current = False

signal_index=2
feature_type=0
nfeatures=12
nepochs=2000
train_instance=4
ncomponents=3
args = Namespace( signal=signal_index, feature_type=feature_type, nfeatures=nfeatures,
	              data_dir="/explore/nobackup/projects/ilab/data/astrotime/demo", dropout_frac=0.5, nepochs=nepochs,
				  batch_size=512, nstreams=10, device=0, loss="mae", learning_rate=0.01 )
# hv.help(hv.Scatter3D)
```

```python
data=tmodel.get_demo_data()
Y, T, E = tmodel.apply_embedding_model( data, args, f"ti{nepochs}_{train_instance}" )
```

```python
E: np.ndarray = tmodel.PCA(E,ncomponents)
manifold: np.ndarray = np.concatenate( (E,Y[:,None]), axis=1)
```

```python
title=f'PCA Feature Manifold, Signal {signal_index}, Feature Type {feature_type}'
manifold_plot = hv.Scatter3D( manifold, kdims=['x', 'y', 'z'], vdims=['value'] ).opts( title=title, size=1, cmap='fire', color='value' )
manifold_plot
```
