```python
from argparse import Namespace
import tmodel
import numpy as np
import pandas as pd
import holoviews as hv
hv.extension('plotly')
use_current = False

signal_index=2
feature_type=0
nfeatures=9
ncomponents=3
args = Namespace( signal=signal_index, feature_type=feature_type, nfeatures=nfeatures,  data_dir="/explore/nobackup/projects/ilab/data/astrotime/demo" )
hv.help(hv.Scatter3D)
```

```python
data=tmodel.get_demo_data()
T: np.ndarray = data['times'][signal_index]
Y: np.ndarray = data['signals'][signal_index]
F: np.ndarray = tmodel.get_features( T, args ).transpose()
```

```python
Y0,Y1,perplexity = Y.min(), Y.max(), 30.0
PC: np.ndarray = tmodel.UMAP(F,ncomponents)
manifold = pd.DataFrame( dict( x=PC[0], y=PC[1], z=PC[2], c=(Y-Y0)/(Y1-Y0) ) )
```

```python
title=f'TSNE Feature Manifold, Signal {signal_index}, Feature Type {feature_type}'
manifold_plot = hv.Scatter3D(manifold, kdims=['x', 'y', 'z'], vdims=['c'] ).opts( title=title, size=50, cmap='fire' )
manifold_plot
```
