```python
from argparse import Namespace
import tmodel
import numpy as np
import holoviews as hv
from holoviews import opts
hv.extension('bokeh')
opts.defaults( opts.Curve(width=1000, height=600, line_width=1) )
use_current = False

signal_index=2
feature_type=0
nfeatures=9
ncomponents=3
args = Namespace( signal=signal_index, feature_type=feature_type, nfeatures=nfeatures,  data_dir="/explore/nobackup/projects/ilab/data/astrotime/demo" )
```

```python
data=tmodel.get_demo_data()
T: np.ndarray = data['times'][signal_index]
Y: np.ndarray = data['signals'][signal_index]
F: np.ndarray = tmodel.get_features( T, args ).transpose()
Y0,Y1 = Y.min(), Y.max()
```

```python
PC: np.ndarray = tmodel.TSNE(F,ncomponents)
colors = (Y-Y0)/(Y1-Y0)
```

```python
manifold = hv.Dataset( (F[0], F[1], F[2], colors), kdims=['x', 'y', 'z'], vdims=['value'] )
title=f'TSNE Feature Manifold, Signal {signal_index}, Feature Type {feature_type}'
opts = hv.opts.Scatter3D( marker='circle', size=1, cmap='viridis', colorbar=True, title=title )
manifold_plot = manifold.to( hv.Scatter3D, vdims=['value'] ).opts( opts )

manifold_plot
```
