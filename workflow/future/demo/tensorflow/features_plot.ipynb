```python
from argparse import Namespace
import tmodel
import numpy as np
import holoviews as hv
from holoviews import opts
hv.extension('bokeh')
opts.defaults(opts.Curve(line_width=1))
use_current = True

if use_current:
    args: Namespace = tmodel.load_args()
    signal_index=args.signal
    feature_type=args.feature_type
else:
    signal_index=2
    feature_type=3
    args: Namespace = tmodel.load_args(signal_index,feature_type)
```

```python
data=tmodel.get_demo_data()
T: np.ndarray = data['times'][signal_index]
F: np.ndarray = tmodel.get_features( T, feature_type, args ).transpose()
```

```python
curve_dict = { fi: hv.Curve((T, F[fi]), 'Time', 'Amplitude')  for fi in range(F.shape[1]) }
kdims = [ hv.Dimension(('feature', 'Feature'), default=0) ]
holomap = hv.HoloMap(curve_dict, kdims=kdims)
holomap.opts(opts.Curve(width=1200))
```
