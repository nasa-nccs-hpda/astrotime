```python
from argparse import Namespace
import tmodel
import numpy as np
import holoviews as hv
from holoviews import opts
hv.extension('bokeh')
opts.defaults( opts.Curve(width=1000, height=600, line_width=1) )
use_current = True

if use_current:
    args: Namespace = tmodel.load_args()
    signal_index=args.signal
    feature_type=args.feature_type
else:
    signal_index=2
    feature_type=3
    args: Namespace = tmodel.load_args(signal_index,feature_type)
```

```python
data=tmodel.get_demo_data()
T: np.ndarray = data['times'][signal_index]
F: np.ndarray = tmodel.get_features( T, feature_type, args ).transpose()
```

```python
curve_dict = { fi: hv.Curve((T, F[fi]), 'Time', 'Amplitude')  for fi in range(F.shape[0]) }
kdims = [ hv.Dimension(('feature', 'Feature'), default=0) ]
holomap = hv.HoloMap(curve_dict, kdims=kdims)
holomap.opts(opts.Curve(width=1200))
```

```python
curve_dict1 = { (fc,fr): hv.Curve((T, F[fr*2+fc]), 'Time', 'Amplitude')  for fr in range(F.shape[0]//2-1,-1,-1) for fc in range(1,-1,-1) }
kdims = [ hv.Dimension(('fr', 'Feature0'), default=0), hv.Dimension(('fc', 'Feature1'), default=0) ]
holomap1 = hv.HoloMap(curve_dict1, kdims=kdims)
grid1 = hv.GridSpace(holomap1)
grid1.opts( opts.GridSpace(plot_size=500))
```
