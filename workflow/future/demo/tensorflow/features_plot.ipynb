```python
import panel.widgets as pnw
import xarray as xa
from argparse import Namespace
import tmodel
import hvplot as hv
import numpy as np
import pandas as pd
import hvplot.pandas  # noqa
import hvplot as hv
hv.extension('bokeh')
import tmodel
use_current = True

if use_current:
    args: Namespace = tmodel.load_args()
    signal_index=args.signal
    feature_type=args.feature_type
else:
    signal_index=2
    feature_type=3
    args: Namespace = tmodel.load_args(signal_index,feature_type)
```

```python
data=tmodel.get_demo_data()
T: np.ndarray = data['times'][signal_index]
F: np.ndarray = tmodel.get_features( T, feature_type, args ).transpose()
```

```python
df = pd.DataFrame({'x': T, 'f': F})
df.hvplot.line( x='x', y='f', title='Features', frame_width=1000, subplots=True ).cols(8)
```
